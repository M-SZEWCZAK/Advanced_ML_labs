```python
class ADABoost:
    def __init__(self,B,base_classifier="tree",max_depth=2):
        self.B=B #number of models in sequence
        self.models=[]
        self.betas=[]
        self.base_classifier=base_classifier
        self.max_depth=max_depth
    @staticmethod
    def resolve_base_classifier(self):
        if self.base_classifier=="tree":
            return DecisionTreeClassifier(max_depth=self.max_depth)
        elif self.base_classifier=="logistic_regression":
            return LogisticRegression()
        else:
            raise ValueError(f"Invalid base classifier: {self.base_classifier}! Valid options are \"tree\" and \"logistic_regression\".")
    def fit(self, X, y):
        weights=np.ones(len(X))/len(X)
        for i in range(self.B):
            f=self.resolve_base_classifier()
            f.fit(X,y,weights)
            y_hat=f.predict(X)
            mask = (y_hat != y)
            w_sel=weights[mask]
            eps=np.sum(w_sel)
            beta=eps/(1-eps)
            self.models.append(f)
            self.betas.append(beta)
            weights[mask]=weights[mask]*beta
            weights=weights/np.sum(weights)
    def predict(self,X):
        pass
```
