```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import sklearn as sk
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from boruta import boruta_py
import scipy
```

```python
def gen1(n,p,k):
    data=np.zeros([n,p])
    y=np.zeros(n)
    chik=scipy.stats.chi2.ppf(0.5, k)
    for i in range(p):
        data[:,i]=np.random.normal(0,1,n)
    for i in range(n):
        subset=data[i,0:k]
        sq=subset**2
        if np.sum(sq)<chik:
            y[i]=0
        else:
            y[i]=1
    return data,y
def gen2(n,p,k):
    data=np.zeros([n,p])
    y=np.zeros(n)
    for i in range(p):
        data[:,i]=np.random.normal(0,1,n)
    for i in range(n):
        subset_sum=np.sum(data[i,0:k])
        if subset_sum>k:
            y[i]=1
        else:
            y[i]=0
    return data,y
def combinex_y(x,y):
    train=pd.concat([x,y],axis=1)
    return train
```
