```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

```python
def generate_artificial(n,b,k,alpha):
    beta=np.zeros(k+5)
    beta[0:5]=b
    x=np.zeros([n,k+5])
    for i in range(n):
        x[i,:]=np.random.normal(0,1,k+5)
    p=np.zeros(n)
    for i in range(n):
        eta=alpha+np.dot(x[i,:],beta)
        p[i]=sigmoid(eta)
    y=np.random.binomial(1,p,n)
    return x,y
```
