```python
def generate_logistic(n,irrelevant_features=10):
    X=[[] for i in range(10+irrelevant_features)]
    for i in range(10+irrelevant_features):
        tmp=np.random.normal(0,1,n)
        X[i]=tmp
    X=np.array(X)
    X=X.T
    sum=np.zeros(n)
    for i in range(10):
        sum=sum+X[:,i]
    p=1/(1+np.exp(-1*(0.5+sum)))
    y=np.random.binomial(1,p,n)
    return X,y
```
